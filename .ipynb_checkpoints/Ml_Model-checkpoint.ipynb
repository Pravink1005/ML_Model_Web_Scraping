```python
import pandas as pd

df = pd.read_csv("/content/drive/MyDrive/Degree_Prediction_Training_5000_Corrected.csv")

print("Dataset loaded successfully!")
print("Number of rows:", len(df))
print("Number of columns:", len(df.columns))

print("\nColumn names:")
print(df.columns.tolist())

df.head()
```

```python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate job descriptions:")
print(df["job_description"].duplicated().sum())

print("\nDegree distribution:")
print(df["degree"].value_counts())

print("\nSpecialization distribution:")
print(df["specialization"].value_counts())
```

```python
cyber_data = df[
    df["job_description"].str.contains(
        "Cyber Security Analyst",
        case=False,
        na=False
    )
]

display(
    cyber_data[
        ["job_description", "degree", "specialization"]
    ].head(10)
)
```

```python
X = df["job_description"]
y = df["degree"]

print("Input data (X):", X.shape)
print("Target data (y):", y.shape)

print("\nExample input:")
print(X.iloc[0])

print("\nExample target:")
print(y.iloc[0])
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("========================================")
print("DATA SPLIT COMPLETED")
print("========================================")

print("Total records:", len(X))
print("Training records:", len(X_train))
print("Testing records:", len(X_test))

print("\nTraining degree distribution:")
print(y_train.value_counts())

print("\nTesting degree distribution:")
print(y_test.value_counts())
```

```python
# Remove duplicate job descriptions

print("Before removing duplicates:", len(df))

df = df.drop_duplicates(
    subset=["job_description"]
).reset_index(drop=True)

print("After removing duplicates:", len(df))

print("\nNew degree distribution:")
print(df["degree"].value_counts())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Create TF-IDF vectorizer
vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    ngram_range=(1, 2),
    max_features=5000
)

# Learn vocabulary from training data
X_train_tfidf = vectorizer.fit_transform(X_train)

# Transform testing data using the same vocabulary
X_test_tfidf = vectorizer.transform(X_test)

print("========================================")
print("TF-IDF CONVERSION COMPLETED")
print("========================================")

print("Training data shape:", X_train_tfidf.shape)
print("Testing data shape:", X_test_tfidf.shape)
print("Number of TF-IDF features:", len(vectorizer.get_feature_names_out()))
```

```python
from sklearn.linear_model import LogisticRegression

# Create the Logistic Regression model
model = LogisticRegression(
    max_iter=2000,
    class_weight="balanced"
)

# Train the model using TF-IDF features
model.fit(X_train_tfidf, y_train)

print("========================================")
print("MODEL TRAINING COMPLETED")
print("========================================")

print("Model:", model)
print("Number of classes:", len(model.classes_))

print("\nDegree classes learned by the model:")
print(model.classes_)
```

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# Predict degree for the unseen test data
y_pred = model.predict(X_test_tfidf)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("========================================")
print("MODEL EVALUATION")
print("========================================")
print("Test Records:", len(y_test))
print("Correct Predictions:", (y_test == y_pred).sum())
print("Wrong Predictions:", (y_test != y_pred).sum())
print("Accuracy:", round(accuracy * 100, 2), "%")

print("\n========================================")
print("CLASSIFICATION REPORT")
print("========================================")

print(
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)
```

```python
# Create a DataFrame containing actual and predicted values

results = pd.DataFrame({
    "job_description": X_test.values,
    "actual_degree": y_test.values,
    "predicted_degree": y_pred
})

# Keep only incorrect predictions
wrong_predictions = results[
    results["actual_degree"] != results["predicted_degree"]
].copy()

print("========================================")
print("WRONG PREDICTIONS")
print("========================================")

print("Total wrong predictions:", len(wrong_predictions))

display(wrong_predictions[
    ["actual_degree", "predicted_degree", "job_description"]
])
```

```python
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# Create confusion matrix
cm = confusion_matrix(y_test, y_pred, labels=model.classes_)

# Plot confusion matrix
plt.figure(figsize=(14, 10))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=model.classes_,
    yticklabels=model.classes_
)

plt.title("Degree Prediction - Confusion Matrix")
plt.xlabel("Predicted Degree")
plt.ylabel("Actual Degree")
plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)
plt.tight_layout()

plt.show()
```

```python
# ========================================
# STEP 12 - NEW JOB PREDICTION
# ========================================

def predict_new_job(job_description):

    # Convert new job description into TF-IDF
    job_tfidf = vectorizer.transform([job_description])

    # Predict degree
    predicted_degree = model.predict(job_tfidf)[0]

    # Get probability
    probabilities = model.predict_proba(job_tfidf)[0]

    # Get confidence
    best_index = probabilities.argmax()
    confidence = probabilities[best_index] * 100

    return predicted_degree, confidence


# ========================================
# TEST 1
# ========================================

job_1 = """
We are hiring a Data Analyst to work with SQL, Excel,
Power BI and business reporting. The candidate will
analyze datasets, create dashboards and provide
insights to business teams. Strong analytical and
communication skills are required.
"""

degree, confidence = predict_new_job(job_1)

print("========================================")
print("NEW JOB 1")
print("========================================")
print("Predicted Degree:", degree)
print("Confidence:", round(confidence, 2), "%")
```

```python
# ========================================
# STEP 13 - CONFIDENCE THRESHOLD TEST
# ========================================

def predict_degree_with_threshold(job_description, threshold=60):

    # Convert job description to TF-IDF
    job_tfidf = vectorizer.transform([job_description])

    # Get probabilities for all degree classes
    probabilities = model.predict_proba(job_tfidf)[0]

    # Find highest probability
    best_index = probabilities.argmax()

    predicted_degree = model.classes_[best_index]
    confidence = probabilities[best_index] * 100

    # Apply threshold
    if confidence >= threshold:
        final_degree = predicted_degree
    else:
        final_degree = "Not Specified"

    return final_degree, confidence


print("✅ Confidence threshold function created!")
```

```python
import re

DEGREE_PATTERNS = {
    "B.E/B.Tech": [
        r"\bB\.?\s*E\.?\b",
        r"\bB\.?\s*Tech\.?\b",
        r"\bBachelor\s+of\s+Engineering\b",
        r"\bBachelor\s+of\s+Technology\b"
    ],

    "B.Sc": [
        r"\bB\.?\s*Sc\.?\b",
        r"\bBachelor\s+of\s+Science\b"
    ],

    "BCA": [
        r"\bBCA\b",
        r"\bBachelor\s+of\s+Computer\s+Applications\b"
    ],

    "B.Com": [
        r"\bB\.?\s*Com\.?\b",
        r"\bBachelor\s+of\s+Commerce\b"
    ],

    "BBA": [
        r"\bBBA\b",
        r"\bBachelor\s+of\s+Business\s+Administration\b"
    ],

    "M.E/M.Tech": [
        r"\bM\.?\s*E\.?\b",
        r"\bM\.?\s*Tech\.?\b",
        r"\bMaster\s+of\s+Engineering\b",
        r"\bMaster\s+of\s+Technology\b"
    ],

    "M.Sc": [
        r"\bM\.?\s*Sc\.?\b",
        r"\bMaster\s+of\s+Science\b"
    ],

    "MCA": [
        r"\bMCA\b",
        r"\bMaster\s+of\s+Computer\s+Applications\b"
    ],

    "MBA": [
        r"\bMBA\b",
        r"\bMaster\s+of\s+Business\s+Administration\b"
    ],

    "PhD": [
        r"\bPh\.?\s*D\.?\b",
        r"\bPhD\b",
        r"\bDoctorate\b"
    ],

    "Diploma": [
        r"\bDiploma\b"
    ],

    "Any Bachelor's Degree": [
        r"\bBachelor'?s\s+degree\b",
        r"\bBachelor'?s\s+qualification\b",
        r"\bAny\s+recognized\s+Bachelor'?s\s+degree\b"
    ],

    "Any Master's Degree": [
        r"\bMaster'?s\s+degree\b",
        r"\bMaster'?s\s+qualification\b",
        r"\bAny\s+recognized\s+Master'?s\s+degree\b"
    ]
}

print("✅ Degree pattern detector created!")
```

```python
def detect_explicit_degree(job_description):

    found_degrees = []

    for degree, patterns in DEGREE_PATTERNS.items():

        for pattern in patterns:

            if re.search(pattern, job_description, re.IGNORECASE):
                found_degrees.append(degree)
                break

    return found_degrees
```

```python
test_text = """
We are looking for a Financial Analyst with an MBA
in Finance. Strong Excel and financial modelling skills
are required.
"""

found = detect_explicit_degree(test_text)

print("Degrees found:", found)
```

```python
# ========================================
# STEP 22 - IMPROVED HYBRID DEGREE MODEL
# ========================================

def hybrid_predict_degree(job_description, ml_threshold=20):

    text = job_description.lower()

    # ----------------------------------------
    # STEP 1: Detect explicitly mentioned degrees
    # ----------------------------------------

    explicit_degrees = detect_explicit_degree(job_description)

    # Remove duplicates while keeping order
    explicit_degrees = list(dict.fromkeys(explicit_degrees))

    # ----------------------------------------
    # STEP 2: Handle Bachelor's + Master's
    # ----------------------------------------

    has_bachelor = (
        "bachelor's" in text
        or "bachelors" in text
        or "bachelor degree" in text
        or "bachelor's degree" in text
    )

    has_master = (
        "master's" in text
        or "masters" in text
        or "master degree" in text
        or "master's degree" in text
    )

    # If both are mentioned
    if has_bachelor and has_master:

        return {
            "degree": "Any Bachelor's Degree / Any Master's Degree",
            "confidence": 100.0,
            "method": "Explicit Degree Detection"
        }

    # ----------------------------------------
    # STEP 3: If exactly one explicit degree
    # ----------------------------------------

    if len(explicit_degrees) == 1:

        return {
            "degree": explicit_degrees[0],
            "confidence": 100.0,
            "method": "Explicit Degree Detection"
        }

    # ----------------------------------------
    # STEP 4: If multiple specific degrees
    # ----------------------------------------

    if len(explicit_degrees) > 1:

        return {
            "degree": " / ".join(explicit_degrees),
            "confidence": 100.0,
            "method": "Explicit Degree Detection"
        }

    # ----------------------------------------
    # STEP 5: No explicit degree → ML
    # ----------------------------------------

    job_tfidf = vectorizer.transform([job_description])

    probabilities = model.predict_proba(job_tfidf)[0]

    best_index = probabilities.argmax()

    predicted_degree = model.classes_[best_index]

    confidence = probabilities[best_index] * 100

    # ----------------------------------------
    # STEP 6: Low confidence handling
    # ----------------------------------------

    if confidence >= ml_threshold:
        final_degree = predicted_degree
    else:
        final_degree = "Not Specified"

    return {
        "degree": final_degree,
        "confidence": round(confidence, 2),
        "method": "Machine Learning"
    }


print("✅ Improved hybrid degree model created!")
```

```python
real_job = """
Senior Data Engineer

Bachelor's or master's degree in computer science,
Data Science, or a related field.

The candidate should have strong expertise in
data engineering, data warehousing, Snowflake,
ETL, Fivetran, dbt, SQL, Python and cloud platforms.
"""

result = hybrid_predict_degree(real_job)

print("========================================")
print("REAL WORLD JOB PREDICTION")
print("========================================")

print("Predicted Degree:", result["degree"])
print("Confidence:", result["confidence"], "%")
print("Method:", result["method"])
```

```python
# ========================================
# STEP 1 - SPECIALIZATION DATA
# ========================================

X_spec = df["job_description"]
y_spec = df["specialization"]

print("========================================")
print("SPECIALIZATION DATA")
print("========================================")

print("Input data (X):", X_spec.shape)
print("Target data (y):", y_spec.shape)

print("Number of classes:", y_spec.nunique())

print("\nSpecialization distribution:")
print(y_spec.value_counts())
```

```python
# ========================================
# STEP 2 - TRAIN TEST SPLIT
# ========================================

from sklearn.model_selection import train_test_split

X_spec_train, X_spec_test, y_spec_train, y_spec_test = train_test_split(
    X_spec,
    y_spec,
    test_size=0.20,
    random_state=42,
    stratify=y_spec
)

print("========================================")
print("SPECIALIZATION DATA SPLIT")
print("========================================")

print("Total records:", len(X_spec))

print("Training records:", len(X_spec_train))

print("Testing records:", len(X_spec_test))
```

```python
# ========================================
# STEP 3 - SPECIALIZATION TF-IDF
# ========================================

from sklearn.feature_extraction.text import TfidfVectorizer

spec_vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    ngram_range=(1, 2),
    max_features=5000
)

X_spec_train_tfidf = spec_vectorizer.fit_transform(
    X_spec_train
)

X_spec_test_tfidf = spec_vectorizer.transform(
    X_spec_test
)

print("========================================")
print("SPECIALIZATION TF-IDF")
print("========================================")

print(
    "Training data shape:",
    X_spec_train_tfidf.shape
)

print(
    "Testing data shape:",
    X_spec_test_tfidf.shape
)

print(
    "Number of features:",
    len(spec_vectorizer.get_feature_names_out())
)
```

```python
# ========================================
# STEP 4 - TRAIN SPECIALIZATION MODEL
# ========================================

from sklearn.linear_model import LogisticRegression

spec_model = LogisticRegression(
    max_iter=2000,
    class_weight="balanced"
)

spec_model.fit(
    X_spec_train_tfidf,
    y_spec_train
)

print("========================================")
print("SPECIALIZATION MODEL TRAINING")
print("========================================")

print("✅ Specialization model trained successfully!")

print(
    "Number of classes:",
    len(spec_model.classes_)
)

print("\nClasses learned by the model:")

print(spec_model.classes_)
```

```python
# ========================================
# STEP 5 - SPECIALIZATION MODEL EVALUATION
# ========================================

from sklearn.metrics import accuracy_score, classification_report

y_spec_pred = spec_model.predict(
    X_spec_test_tfidf
)

accuracy = accuracy_score(
    y_spec_test,
    y_spec_pred
)

print("========================================")
print("SPECIALIZATION MODEL EVALUATION")
print("========================================")

print("Test Records:", len(y_spec_test))

print(
    "Correct Predictions:",
    (y_spec_test == y_spec_pred).sum()
)

print(
    "Wrong Predictions:",
    (y_spec_test != y_spec_pred).sum()
)

print(
    "Accuracy:",
    round(accuracy * 100, 2),
    "%"
)

print("\n========================================")
print("CLASSIFICATION REPORT")
print("========================================")

print(
    classification_report(
        y_spec_test,
        y_spec_pred,
        zero_division=0
    )
)
```
